```python
import psycopg2
import pandas as pd
```

```python
from secret_copy import PASSWORD, USER, HOST, PORT
```

```python
# Database name
database = "postgres"

# Construct the connection string
conn_string = f"host={HOST} dbname={database} user={USER} password={PASSWORD} port={PORT}"
```

```python
# Establish the connection and load data into a DataFrame
try:
    conn = psycopg2.connect(conn_string)
    print("Connected to the database successfully!")

    # Create a cursor object to execute SQL queries
    cur = conn.cursor()

    # Execute your SQL query
    cur.execute("SELECT * FROM public.dishes")

    # Fetch all rows from the executed query
    rows = cur.fetchall()

    # Get column names from the cursor description
    columns = [desc[0] for desc in cur.description]

    # Create a DataFrame from the fetched data
    df_menu = pd.DataFrame(rows, columns=columns)

    # Print the DataFrame
    print(df_menu)

    # Close the cursor and connection
    cur.close()
    conn.close()
    print("Database connection closed.")

except (Exception, psycopg2.Error) as error:
    print("Error while connecting to PostgreSQL:", error)
```

```python
# Establish the connection and load data into a DataFrame
try:
    conn = psycopg2.connect(conn_string)
    print("Connected to the database successfully!")

    # Create a cursor object to execute SQL queries
    cur = conn.cursor()

    # Execute SQL query to get unique values from the meats column
    cur.execute("SELECT DISTINCT meats FROM public.dishes WHERE meats IS NOT NULL")

    # Fetch all rows from the executed query
    rows = cur.fetchall()

    # Extract the unique meat values from the result
    unique_meats = [row[0] for row in rows]

    # Print the unique meat values
    print("Unique values in dishes.meats:")
    for meat in unique_meats:
        print(meat)

    # Close the cursor and connection
    cur.close()
    conn.close()
    print("Database connection closed.")

except (Exception, psycopg2.Error) as error:
    print("Error while connecting to PostgreSQL:", error)
```
