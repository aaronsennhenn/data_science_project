##### Load packages

```python
import pandas as pd
import requests
import json
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
import transformers
import os
from lxml import etree
import sqlite3
import re
```

##### Diffusion Model

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Replace the model version with your required version if needed
pipeline = StableDiffusionPipeline.from_pretrained(
    "stabilityai/stable-diffusion-2-1", torch_dtype=torch.float16
)

# Running the inference on GPU with cuda enabled
pipeline = pipeline.to('cuda')
```

##### Image generation

1) Prinz Karl

```python
url = 'https://www.my-stuwe.de//wp-json/mealplans/v1/canteens/623?lang=de&v=1729614874306'

# Make the GET request
response = requests.get(url)

# Parse the JSON response
json_data = response.json()

# Pretty print the JSON response
pretty_json = json.dumps(json_data, indent=4)
print(pretty_json)

# Create and print df
df = pd.json_normalize(json_data)

# Extract the list of dictionaries from the column
menus_list = df['623.menus'].iloc[0]

# Convert the list of dictionaries into a new DataFrame
menus_df = pd.DataFrame(menus_list)

# Display the new DataFrame
print(menus_df)

# Save to excel
menus_df.to_excel('menus_cafeteria_623.xlsx', index=False)

# Create the 'images_prinzkarl' folder if it doesn't exist
if not os.path.exists('images_cafeteria_623'):
    os.makedirs('images_cafeteria_623')

# Iterate over each row in the DataFrame
for index, row in menus_df.iterrows():
    # Set up the prompt
    prompt = row["menu"]
    
    try:
        # Generate the image
        image = pipeline(prompt=prompt).images[0]
        
        # Define the filename using the 'id'
        filename = f"{row['id']}_image.png"
        
        # Create the full path for saving the image
        filepath = os.path.join('images_cafeteria_623', filename)
        
        # Save the generated image
        image.save(filepath)
        
        # Append the filename to a new column in the DataFrame
        menus_df.at[index, 'image_filename'] = filename
        print(f"Image saved: {filepath}")
    except Exception as e:
        print(f"Error processing row {index}: {str(e)}")
        menus_df.at[index, 'image_filename'] = "error"

# Display the updated DataFrame
print(menus_df)
```

##### Data acquisition and storage

```python
def process_url(url):
    response = requests.get(url)
    json_data = response.json()
    
    canteen_number = re.search(r'canteens/(\d+)', url).group(1)
    
    menus_data = json_data[canteen_number]['menus']
    df_menus = pd.json_normalize(menus_data)
    
    columns_to_remove = ['photo.thumbnail', 'photo.medium', 'photo.large', 'photo.full']
    df_menus = df_menus.drop(columns=columns_to_remove)
    
    # Convert complex data types to strings
    for col in df_menus.columns:
        if df_menus[col].dtype == 'object':
            df_menus[col] = df_menus[col].apply(lambda x: json.dumps(x) if isinstance(x, (list, dict)) else x)
    
    return canteen_number, df_menus

def table_exists(conn, table_name):
    cursor = conn.cursor()
    cursor.execute(f"SELECT name FROM sqlite_master WHERE type='table' AND name='{table_name}';")
    return cursor.fetchone() is not None
```

```python
urls = [
    'https://www.my-stuwe.de//wp-json/mealplans/v1/canteens/623?lang=de&v=1729614874306',
    'https://www.my-stuwe.de//wp-json/mealplans/v1/canteens/715?lang=de&v=1729619513869',
    'https://www.my-stuwe.de//wp-json/mealplans/v1/canteens/724?lang=de&v=1729621051545',
    'https://www.my-stuwe.de//wp-json/mealplans/v1/canteens/621?lang=de&v=1729706358572',
    'https://www.my-stuwe.de//wp-json/mealplans/v1/canteens/611?lang=de&v=1729706211893'
]

# Connect to SQLite database
conn = sqlite3.connect('menu_data.db')

for url in urls:
    canteen_number, df = process_url(url)
    
    table_name = f'canteen_{canteen_number}'
    
    if table_exists(conn, table_name):
        print(f"Table {table_name} already exists. Skipping creation.")
    else:
        # Define data types for SQLite
        dtypes = {col_name: 'TEXT' for col_name in df.columns}
        
        # Store the DataFrame in the SQLite database
        df.to_sql(table_name, conn, if_exists='replace', index=False, dtype=dtypes)
        print(f"Data for canteen {canteen_number} has been stored in table {table_name}")

conn.close()
```

## Data analysis

```python
# Connect to the database
db_path = os.path.join(os.getcwd(), 'menu_data.db')
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# List of canteen tables
canteen_tables = ['canteen_623', 'canteen_715', 'canteen_724', 'canteen_621', 'canteen_611']

for table in canteen_tables:
    print(f"\nTable: {table}")
    print("=" * 80)

    # Get column names
    cursor.execute(f"PRAGMA table_info({table})")
    columns = [column[1] for column in cursor.fetchall()]

    # Print column names
    print(" | ".join(columns))
    print("-" * 80)

    # Get first 5 rows
    cursor.execute(f"SELECT * FROM {table} LIMIT 5")
    rows = cursor.fetchall()

    # Print rows
    for row in rows:
        print(" | ".join(str(value) for value in row))

    print("\n")

conn.close()
```
